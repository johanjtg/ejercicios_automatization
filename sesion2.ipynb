## Ejercicio 1 — Analizador de portafolio basico

Almacena un portafolio en un diccionario, calcula el valor total e identifica la accion con mayor inversion.

```python
portafolio = {
    'AAPL':  {'cantidad': 100, 'precio_compra': 150},
    'GOOGL': {'cantidad': 50,  'precio_compra': 130},
    'MSFT':  {'cantidad': 75,  'precio_compra': 300}
}

# Calcular valor total de cada accion
for ticker, datos in portafolio.items():
    datos['valor_total'] = datos['cantidad'] * datos['precio_compra']

valor_total_portafolio = sum(d['valor_total'] for d in portafolio.values())

# Identificar la accion con mayor inversion
accion_mayor = max(portafolio, key=lambda t: portafolio[t]['valor_total'])

# Mostrar resultados
print('=== Portafolio ===')
for ticker, datos in portafolio.items():
    print(f"  {ticker}: {datos['cantidad']} acciones x ${datos['precio_compra']} = ${datos['valor_total']:,}")

print(f"\nValor total del portafolio: ${valor_total_portafolio:,}")
print(f"Accion con mayor inversion: {accion_mayor} (${portafolio[accion_mayor]['valor_total']:,})")
```

## Ejercicio 2 — Filtrador de acciones

Funcion que recibe una lista de diccionarios con acciones, filtra las que tienen precio > $100, las ordena por precio descendente y retorna solo los tickers.

```python
acciones = [
    {'ticker': 'AAPL', 'precio': 175.50},
    {'ticker': 'NVDA', 'precio': 880.00},
    {'ticker': 'AMD',  'precio': 95.00},
    {'ticker': 'MSFT', 'precio': 420.00},
    {'ticker': 'INTC', 'precio': 45.00}
]

def filtrar_acciones_caras(acciones, precio_minimo=100):
    filtradas = [a for a in acciones if a['precio'] > precio_minimo]
    ordenadas = sorted(filtradas, key=lambda a: a['precio'], reverse=True)
    return [a['ticker'] for a in ordenadas]

resultado = filtrar_acciones_caras(acciones)
print(resultado)  # ['NVDA', 'MSFT', 'AAPL']
```
